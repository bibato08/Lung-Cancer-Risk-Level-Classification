## Feature Engineering

```python
#Environmental Risk Score

df_clean["Environmental_Risk"] = (
    df_clean["Air Pollution"] +
    df_clean["Dust Allergy"] +
    df_clean["OccuPational Hazards"]
)
```

```python
#Lifestyle Risk Score

df_clean["Lifestyle_Risk"] = (
    df_clean["Smoking"] +
    df_clean["Passive Smoker"] +
    df_clean["Alcohol use"] +
    df_clean["Obesity"]
)
```

```python
#Respiratory Symptom Score

df_clean["Respiratory_Symptom_Score"] = (
    df_clean["Coughing of Blood"] +
    df_clean["Shortness of Breath"] +
    df_clean["Wheezing"] +
    df_clean["Dry Cough"] +
    df_clean["Frequent Cold"]
)
```

```python
#General Health Risk Score
df_clean["Health_Risk_Score"] = (
    df_clean["Fatigue"] +
    df_clean["Weight Loss"] +
    df_clean["Chest Pain"] +
    df_clean["Swallowing Difficulty"] +
    df_clean["Clubbing of Finger Nails"]
)
```

```python
#Overall Risk Score
df_clean["Overall_Risk_Score"] = (
    df_clean["Environmental_Risk"] +
    df_clean["Lifestyle_Risk"] +
    df_clean["Respiratory_Symptom_Score"] +
    df_clean["Health_Risk_Score"]
)
```

```python
X = df_clean.drop("Level", axis=1)
y = df_clean["Level"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Training set:")
print(y_train.value_counts())

print("\nTesting set:")
print(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

```python
print("Training shape:", X_train_scaled.shape)
print("Testing shape:", X_test_scaled.shape)
```

```python
print(X_train_scaled[:5])
```

```python
from sklearn.linear_model import LogisticRegression
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)
```

```python
logistic_model.fit(X_train_scaled, y_train)
```

```python
y_pred_logistic = logistic_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
accuracy = accuracy_score(y_test, y_pred_logistic)

print("Logistic Regression Accuracy:", accuracy)
```

```python
print(classification_report(y_test, y_pred_logistic))
```

```python
cm = confusion_matrix(y_test, y_pred_logistic)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=logistic_model.classes_,
    yticklabels=logistic_model.classes_
)

plt.title("Logistic Regression Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
#5-Fold Stratified Cross-Validation

from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
#Create 5-fold stratified CV

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
#Logistic Regression CV

logistic_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000, random_state=42))
])

logistic_cv = cross_val_score(
    logistic_pipeline,
    X_original,
    y,
    cv=cv,
    scoring="accuracy"
)

print("Logistic Regression CV Scores:", logistic_cv)
print("Mean Accuracy:", logistic_cv.mean())
print("Standard Deviation:", logistic_cv.std())
```

```python
#KNN CV

knn_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier(n_neighbors=5))
])

knn_cv = cross_val_score(
    knn_pipeline,
    X_original,
    y,
    cv=cv,
    scoring="accuracy"
)

print("KNN CV Scores:", knn_cv)
print("Mean Accuracy:", knn_cv.mean())
print("Standard Deviation:", knn_cv.std())
```

```python
#Decision Tree CV
dt_cv = cross_val_score(
    DecisionTreeClassifier(random_state=42),
    X_original,
    y,
    cv=cv,
    scoring="accuracy"
)

print("Decision Tree CV Scores:", dt_cv)
print("Mean Accuracy:", dt_cv.mean())
print("Standard Deviation:", dt_cv.std())
```

```python
#Random Forest CV

rf_cv = cross_val_score(
    RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ),
    X_original,
    y,
    cv=cv,
    scoring="accuracy"
)

print("Random Forest CV Scores:", rf_cv)
print("Mean Accuracy:", rf_cv.mean())
print("Standard Deviation:", rf_cv.std())
```

```python
#SVM CV
svm_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", SVC(
        kernel="rbf",
        C=1.0,
        random_state=42
    ))
])

svm_cv = cross_val_score(
    svm_pipeline,
    X_original,
    y,
    cv=cv,
    scoring="accuracy"
)

print("SVM CV Scores:", svm_cv)
print("Mean Accuracy:", svm_cv.mean())
print("Standard Deviation:", svm_cv.std())
```
